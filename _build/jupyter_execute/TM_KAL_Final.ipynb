Membuat program matplotlib python agar (ABCD) dapat bergerak ke (IJKL), kemudian dari (IJKL) menuju ke titik (0,0). Hal yang sama juga dilakukan pada cerminannya.

## Hasil:



```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# --- KONFIGURASI ---
steps_per_phase = 60  # 60 frame untuk turun, 60 frame untuk menciut
total_frames = steps_per_phase * 2

def get_rect_coords(x_range, y_range):
    return [x_range[0], x_range[1], x_range[1], x_range[0], x_range[0]], \
           [y_range[1], y_range[1], y_range[0], y_range[0], y_range[1]]

# Definisi Titik Y
y_abcd = (3, 4)
y_ijkl = (1, 2)
y_zero = (0, 0) # Titik tujuan akhir (berimpit)

fig, ax = plt.subplots(figsize=(7, 7))
ax.set_aspect('equal', adjustable='box')

ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_xticks(range(-5, 6))
ax.set_yticks(range(-5, 6))
ax.grid(True, which='both', color='lightgray', linestyle='-', linewidth=0.5)
ax.set_axisbelow(True)
ax.axhline(0, color='black', lw=2)
ax.axvline(0, color='black', lw=1.5)

line_main, = ax.plot([], [], 'b-o', lw=2, markersize=8, label='Blok Utama')
line_mirror, = ax.plot([], [], 'r-o', lw=2, markersize=8, label='Blok Cermin')
ax.legend(loc='upper right')

is_paused = False

def init():
    line_main.set_data([], [])
    line_mirror.set_data([], [])
    return line_main, line_mirror

def update(frame):
    if is_paused:
        return line_main, line_mirror

    if frame < steps_per_phase:
        # --- FASE 1: ABCD ke IJKL (Tetap Persegi) ---
        t = frame / float(steps_per_phase)
        curr_low = y_abcd[0] + (y_ijkl[0] - y_abcd[0]) * t
        curr_high = y_abcd[1] + (y_ijkl[1] - y_abcd[1]) * t
    else:
        # --- FASE 2: IJKL ke 0 (Menciut) ---
        t = (frame - steps_per_phase) / float(steps_per_phase)
        # Sisi bawah (1) bergerak ke 0, sisi atas (2) juga bergerak ke 0
        curr_low = y_ijkl[0] + (y_zero[0] - y_ijkl[0]) * t
        curr_high = y_ijkl[1] + (y_zero[1] - y_ijkl[1]) * t

    x_vals, y_vals = get_rect_coords([2, 3], [curr_low, curr_high])

    line_main.set_data(x_vals, y_vals)
    line_mirror.set_data(x_vals, [-y for y in y_vals])

    return line_main, line_mirror

def toggle_pause(event):
    global is_paused
    is_paused = not is_paused
    ax.set_title("PAUSED" if is_paused else "Play")
    fig.canvas.draw_idle()

fig.canvas.mpl_connect('button_press_event', toggle_pause)

ani = FuncAnimation(fig, update, frames=total_frames,
                    init_func=init, blit=True, interval=25, repeat=True)

plt.title("ABCD bergerak ke IJKL, dan IJKL bergerak ke titik (0,0) Begitu juga dengan cerminannya")
plt.tight_layout()

# Use HTML5 video rendering for Jupyter/GitHub friendliness
plt.close()
from IPython.display import HTML
plt.close() # Menghindari muncul gambar double
HTML(ani.to_jshtml())
```
